```python


import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import joblib

SAVE_DIR = "/content/drive/MyDrive/road_accident_model"
os.makedirs(SAVE_DIR, exist_ok=True)
print("All files will be saved to:", SAVE_DIR)

np.random.seed(42)
n = 1200
speed_limit = np.random.choice([30,50,60,80,100,120], size=n, p=[0.15,0.25,0.20,0.20,0.15,0.05])
actual_speed = speed_limit + np.random.normal(loc=0, scale=8, size=n)
weather = np.random.choice([0,1,2], size=n, p=[0.7,0.2,0.1])
road_surface = np.random.choice([0,1], size=n, p=[0.85,0.15])
lighting = np.random.choice([0,1], size=n, p=[0.75,0.25])
driver_age = np.random.normal(35,12,size=n).clip(16,90).astype(int)
seatbelt = np.random.choice([0,1], size=n, p=[0.12,0.88])
alcohol = np.random.choice([0,1], size=n, p=[0.92,0.08])
vehicle_type = np.random.choice([0,1,2], size=n, p=[0.75,0.15,0.10])
num_vehicles = np.random.choice([1,2,3,4], size=n, p=[0.6,0.25,0.1,0.05])
traffic_density = np.random.uniform(0.1,1.0,size=n)

severity = (
    0.03*actual_speed + 0.8*lighting + 1.1*(weather==1)
    + 1.6*(weather==2) + 1.4*road_surface + 1.8*alcohol
    + 0.9*(vehicle_type==1) - 0.01*driver_age - 1.2*seatbelt
    + 0.4*num_vehicles + 0.5*traffic_density*3.0
    + np.random.normal(scale=1.2, size=n)
)
sev_min, sev_max = severity.min(), severity.max()
severity_scaled = np.clip(10*(severity-sev_min)/(sev_max-sev_min), 0, 10)

df = pd.DataFrame({
    "speed_limit": speed_limit,
    "actual_speed": np.round(actual_speed,1),
    "weather": weather,
    "road_surface_wet": road_surface,
    "lighting_night": lighting,
    "driver_age": driver_age,
    "seatbelt_on": seatbelt,
    "alcohol": alcohol,
    "vehicle_type": vehicle_type,
    "num_vehicles": num_vehicles,
    "traffic_density": np.round(traffic_density,2),
    "severity_score": np.round(severity_scaled,3)
})
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

csv_path = os.path.join(SAVE_DIR, "road_accident_synthetic.csv")
df.to_csv(csv_path, index=False)
print("Dataset saved at:", csv_path)

X = df.drop(columns=["severity_score"])
y = df["severity_score"]
X_encoded = pd.get_dummies(X, columns=["weather","vehicle_type"], drop_first=True)

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.25, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)
r2_train = r2_score(y_train, y_pred_train)
r2_test = r2_score(y_test, y_pred_test)
mse_test = mean_squared_error(y_test, y_pred_test)

print("Train R2:", round(r2_train,4))
print("Test R2:", round(r2_test,4))
print("Test MSE:", round(mse_test,4))

model_path = os.path.join(SAVE_DIR, "linear_regression_accident_model.joblib")
joblib.dump({"model": model, "columns": list(X_train.columns)}, model_path)
print("Model saved at:", model_path)

demo_script = """bundle = joblib.load("linear_regression_accident_model.joblib")
model = bundle["model"]
cols = bundle["columns"]

example = {
    "speed_limit": 50,
    "actual_speed": 78.0,
    "road_surface_wet": 1,
    "lighting_night": 1,
    "driver_age": 28,
    "seatbelt_on": 0,
    "alcohol": 1,
    "num_vehicles": 2,
    "traffic_density": 0.7,
}

df_new = pd.DataFrame([example])
for c in cols:
    if c not in df_new.columns:
        df_new[c] = 0
df_new = df_new[cols]
print("Predicted severity (0-10):", model.predict(df_new)[0])
"""
with open(os.path.join(SAVE_DIR, "demo_load_model.py"), "w") as f:
    f.write(demo_script)

with open(os.path.join(SAVE_DIR, "this_readme.txt"), "w") as f:
    f.write("Files generated: road_accident_synthetic.csv, linear_regression_accident_model.joblib, demo_load_model.py\n")

hypo = {
    "speed_limit": 50,
    "actual_speed": 78.0,
    "road_surface_wet": 1,
    "lighting_night": 1,
    "driver_age": 28,
    "seatbelt_on": 0,
    "alcohol": 1,
    "num_vehicles": 2,
    "traffic_density": 0.7,
}
hypo_df = pd.DataFrame([hypo])
for col in X_train.columns:
    if col not in hypo_df.columns:
        hypo_df[col] = 0
hypo_df = hypo_df[X_train.columns]
pred = model.predict(hypo_df)[0]

print("Predicted Accident Severity Score (0-10):", round(pred,3))

print("Dataset preview")
print(df.head(10))
```
